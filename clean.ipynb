```python
%load_ext autoreload
%autoreload 2

from scripts.plot_results import *

# Default configuration of parameters
config_default = {'NUM_SITES': 12, 'NUM_ELECTRONS': 12, 'N_UP': 6, 'N_DN': 6, 'T1': 1.0, 'T2': 0.2, 'PHI': 0.5, 'TOL': 1e-9, 'N_FLUX': 4}
```

```python
tol_list = [1e-6, 1e-7, 1e-8, 1e-9]
config_list = []
for tol in tol_list:
    config_list.append(config_default.copy())
    config_list[-1]['TOL'] = tol
    points = get_points(config_list[-1])
    print(points)

labels = [f'$\eta_{{tol}}=10^{{{int(np.log10(tol))}}}$' for tol in tol_list]

fig_ax = plot_1d_cut(config_list, 'u', {'delta': 0.0, 'v': 0.0}, 
                     ['sdw', 'cdw', 'gap'], [r'$S_{SDW}$', r'$S_{CDW}$', r'$\epsilon_{gap}$'], 
                     r'$U/t$', rf'$\Delta/t=0,\, V/t=0$', labels=labels, xlim=(7.95, 8.3))

plt.show()
```

```python
n_flux_list = [4, 6, 8, 10, 12]
config_list = []
for n_flux in n_flux_list:
    config_list.append(config_default.copy())
    config_list[-1]['N_FLUX'] = n_flux
    points = get_points(config_list[-1])
    print(points)

labels = [rf'$N_{{\phi}}={n_flux**2}$' for n_flux in n_flux_list]

fig_ax = plot_1d_cut(config_list, 'u', {'delta': 0.0, 'v': 0.0}, 'chern', r'$\mathcal{C}$', 
                     r'$U/t$', rf'$\Delta/t=0, V/t=0$', labels=labels, xlim=(6.75, 8.25))

plt.show()
```
